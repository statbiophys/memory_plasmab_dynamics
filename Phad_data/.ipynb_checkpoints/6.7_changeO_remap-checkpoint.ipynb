```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
metadata = pd.read_csv('metadata/metadata.tsv', sep='\t', index_col=0)
```

```python
pat = '2'
f_fam = pd.read_csv('lineages/changeO_out/pat' + pat + '_changeO_clone-pass.tsv', sep='\t', index_col='sequence_id')
f_fam.index = f_fam.index.str.split('-').str[0] + '-' + f_fam.index.str.split('-').str[1]
```

```python
for samp, row in metadata[metadata.patient == int(pat)].iterrows():
    f = pd.read_csv('sequences/'+samp+'.tsv', sep='\t', index_col=0, low_memory=False)
    print(samp)
    
    # Mapping families of heavy sequences
    f['family_changeO'] = f.pat_heavy_id.map(f_fam.clone_id)
    fh = f[f.chain == 'H']
    fh = fh[fh.paired_seq.notna()]
    print('check percentage assigned heavy families', np.sum(fh.family_changeO.notna()) / len(fh) * 100)
    
    f.to_csv('sequences/'+samp+'.tsv', sep='\t')
```

```python
for samp, row in metadata[metadata.patient == int(pat)].iterrows():
#for samp in ['pat1_t2_mc']:
    
    f_samp = pd.read_csv('sequences/'+samp+'.tsv', sep='\t', index_col=0, low_memory=False)
    print(samp)
    
    for r in range(metadata.loc[samp, 'repl_count']):
        
        f_r = pd.read_csv('sequences/replicates/'+samp+'_r'+str(r+1)+'.tsv', sep='\t', index_col=0, low_memory=False)
        f_r['family_changeO'] = f_r.sample_id.map(f_samp.family_changeO)
        fh = f_r[f_r.chain == 'H']
        fh = fh[fh.paired_seq.notna()]
        print('check percentage assigned heavy families', np.sum(fh.family_changeO.notna()) / len(fh) * 100)
    
        f_r.to_csv('sequences/replicates/'+samp+'_r'+str(r+1)+'.tsv', sep='\t')
```
